```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Phase boundary model
# -----------------------------
def boundary(x, s, b, c, p):
    return s * x**(-b) * np.exp(c * x * np.exp(-p * x))

# -----------------------------
# Temperature-dependent parameters
# -----------------------------
def b_T(T, b0=0.5, alpha=-0.000, T0=298):
    """
    Tie-line slope ratio b(T)=k2/k1.
    alpha < 0 means b decreases with temperature.
    """
    return b0 * np.exp(alpha * (T - T0))

def epsilon_T(T, eps0=-0.1, deps_dT=0.0, T0=298):
    """
    Effective binding/free-energy scale in units of k_B*T0.
    Here we set k_B = 1, so epsilon has units of Kelvin.
    """
    return eps0 + deps_dT * (T - T0)

def c_T(T, c0=10, eps0=-0.1, deps_dT=0.0):
    eps = epsilon_T(T, eps0=eps0, deps_dT=deps_dT)
    return c0 * (1 / T) * np.exp(-eps / T)

def p_T(T, p0=0.01):
    return p0 * (1 / T)

def bp_over_c(T, b0=0.5, alpha=-0.003, c0=3000, p0=300, eps0=2.0):
    b = b_T(T, b0=b0, alpha=alpha)
    return b * p0 / c0 * np.exp(eps0 / T)

# -----------------------------
# Plot phase boundaries
# -----------------------------
s = 1.0
x = np.linspace(0.05, 100, 1000)

temperatures = [248, 298, 348]  # K

plt.figure(figsize=(5.2, 4.2))

for T in temperatures:
    b = b_T(T)
    c = c_T(T)
    p = p_T(T)

    y = boundary(x, s, b, c, p)

    plt.plot(x, y, label=f"T={T} K, b={b:.2f}")

#plt.yscale("log")
plt.xlabel("RNA concentration x")
plt.ylabel("G3BP1 concentration y")
plt.title("Temperature-dependent phase boundary")
plt.legend(frameon=False, fontsize=8)
plt.tight_layout()
plt.ylim(0,3)
plt.show()

# -----------------------------
# Plot reentrance-strength measure bp/c
# -----------------------------
T_grid = np.linspace(270, 340, 300)
R = bp_over_c(T_grid)

plt.figure(figsize=(5.2, 4.0))
plt.plot(T_grid, R)
plt.axhline(0.16, linestyle="--", label=r"$bp/c=0.16$")
plt.xlabel("Temperature T / K")
plt.ylabel(r"$bp/c$")
plt.title("Reentrance-strength measure")
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```

```python

# ============================================================
# MODEL
# ============================================================

def boundary(x, s, b, c, p):
    return s * x**(-b) * np.exp(c * x * np.exp(-p * x))

def b_T(T):
    return b0 * np.exp(alpha * (T - T0))

def epsilon_T(T):
    return eps0 + deps_dT * (T - T0)

def c_T(T):
    eps = epsilon_T(T)
    return c0 * (1/T) * np.exp(-eps/T)

def p_T(T):
    return p0 * (1/T)

def bp_over_c(T):
    b = b_T(T)
    return b * p0 / c0 * np.exp(epsilon_T(T)/T)
```

```python

# ============================================================
# USER PARAMETERS (easy to tune)
# ============================================================

# phase-boundary prefactor
s = 1.0

# b(T) = b0 exp[alpha(T-T0)]
b0 = 0.5
alpha = 0.0

# UCST-like
#eps0 = +100

# weak temperature dependence
#eps0 = 0

# LCST-like
#eps0 = -100

# epsilon(T) = eps0 + deps_dT(T-T0)
eps0 = 100
deps_dT = 0.0

# c(T) = c0/T * exp[-epsilon(T)/T]
c0 = 10.0

# p(T) = p0/T
p0 = 0.01

# reference temperature
T0 = 298

# temperatures to plot
temperatures = [248, 298, 348]

# plotting ranges
x_min = 0.05
x_max = 100

T_plot_min = 240
T_plot_max = 360


# ============================================================
# DATA
# ============================================================

x = np.linspace(x_min, x_max, 2000)
T_grid = np.linspace(T_plot_min, T_plot_max, 500)

# ============================================================
# FIGURE
# ============================================================

fig, (ax1, ax2) = plt.subplots(
    1, 2,
    figsize=(10,4)
)

# ------------------------------------------------------------
# Left panel: phase boundaries
# ------------------------------------------------------------

for T in temperatures:

    b = b_T(T)
    c = c_T(T)
    p = p_T(T)

    y = boundary(x, s, b, c, p)

    ax1.plot(
        x,
        y,
        lw=2,
        label=fr"$T={T}\,$K"
    )

ax1.set_xlabel("RNA concentration")
ax1.set_ylabel("G3BP1 concentration")
ax1.set_title("Phase boundary")

ax1.set_ylim(0, 3)
ax1.legend(frameon=False)

# ------------------------------------------------------------
# Right panel: reentrance measure
# ------------------------------------------------------------

R = bp_over_c(T_grid)

ax2.plot(
    T_grid,
    R,
    lw=2
)

ax2.axhline(
    0.16,
    ls="--",
    color="k",
    label=r"$bp/c=0.16$"
)

ax2.set_xlabel(r"Temperature $T$")
ax2.set_ylabel(r"$bp/c$")
ax2.set_title("Reentrance strength")

ax2.legend(frameon=False)

# ------------------------------------------------------------
# Layout
# ------------------------------------------------------------

plt.tight_layout()
plt.show()
```
